## Analyzing Dataset vgsales

#### <b>What is the highest sold genre?</b>

Let's first look at the total sales per region. 
Remeber our question: <b>Which region sells the most video games?</b>

```python
sales_totals = vgsales[sales_region].sum()
print(sales_totals)
```

```python
sales_totals_df = sales_totals.reset_index()
sales_totals_df.columns = ['Sales Column', 'Total Sales']
display(sales_totals_df)
```

We can see that NA_sales, which is North America, sells the most video games in total.
Now let's take a look at sales grouped by genere per region.

```python
sales_by_genre = vgsales.groupby('Genre')[sales_region].sum()
display(sales_by_genre)
```

```python
sales_by_genre_sorted = sales_by_genre.sort_values(by='Global_Sales', ascending=False)
display(sales_by_genre_sorted)
```

Let's vizualize the data above to get a better idea what we are looking at. We will use a stacked bar chart for vizualization.

```python
regions_only = sales_by_genre_sorted.drop(columns='Global_Sales')
ax = regions_only.plot(kind='bar', stacked=True, figsize=(12, 7))

ax.set_ylabel('Total Sales in Millions')
ax.set_title('Total Regional Sales by Genre')
ax.legend(title='Region')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

From the above vizualization we can discover a few things. 
<ul>
    <li>The highest sold genre is action.</li>
    <li>Stragety is the lowest sold genre.</li>
    <li>North America and Europe seem to sell more video games than other regions. Or at least what is provided in this dataset.</li>
</ul>

#### Let's take a look at each region individually to get a better idea how a region may differ on their highest and lowest sold genre.

```python
regions = ['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']
colors = ['blue', 'orange', 'green', 'red']  # colors in order

for region, color in zip(regions, colors):
    sorted_region = sales_by_genre.sort_values(by=region, ascending=False).reset_index()
    
    plt.figure(figsize=(10, 5))
    plt.bar(sorted_region['Genre'], sorted_region[region], color=color)
    plt.title(f'Sales by Genre in {region}')
    plt.ylabel('Sales (Millions)')
    plt.xlabel('Genre')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
```

We can see that action, sports, and shooters are the top three genres in each region except Japan. Japan's top three sold genres are role-playing, action, and sports; where their lowest sold genre is shooter.  

#### Is there a correlation between publishers and video game sales?

```python
publisher_sales = vgsales.groupby('Publisher')['Global_Sales'].sum().sort_values(ascending=False)
publisher_sales.head(10)
```

```python
top_publishers = publisher_sales.head(10)

plt.figure(figsize=(10, 5))
top_publishers.plot(kind='bar', color='skyblue')
plt.title('Top 10 Publishers by Global Sales')
plt.ylabel('Global Sales (Millions)')
plt.xlabel('Publisher')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
publisher_dummies = pd.get_dummies(vgsales['Publisher'])

# Add sales column
publisher_dummies['Global_Sales'] = vgsales['Global_Sales']

# Compute correlations
publisher_corr = publisher_dummies.corr()['Global_Sales'].sort_values(ascending=False)

publisher_corr.head(10)
```

```python
avg_sales_by_publisher = vgsales.groupby('Publisher')['Global_Sales'].mean().sort_values(ascending=False)
avg_sales_by_publisher.head(10)
```
