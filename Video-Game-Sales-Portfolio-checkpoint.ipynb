Let's try scraping the internet to see if we can fill in the missing data.

```python
# Sales columns in your dataset
sales_cols = ['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales']

def get_game_info(game_name):
    info = {"Year": None, "Publisher": None}
    for col in sales_cols:
        info[col] = None  # initialize sales columns

    try:
        # --- Wikipedia scraping for metadata ---
        url = f"https://en.wikipedia.org/wiki/{game_name.replace(' ', '_')}"
        response = requests.get(url, timeout=10)
        if response.status_code == 200:
            soup = BeautifulSoup(response.text, 'html.parser')
            infobox = soup.find('table', {'class': 'infobox'})
            if infobox:
                for row in infobox.find_all('tr'):
                    header = row.find('th')
                    data = row.find('td')
                    if header and data:
                        header_text = header.text.strip().lower()

                        if 'publisher' in header_text:
                            info['Publisher'] = data.text.strip()

                        elif 'release' in header_text or 'released' in header_text:
                            match = re.search(r"\b(19|20)\d{2}\b", data.text)
                            if match:
                                info['Year'] = int(match.group())

        # --- VGChartz scraping for sales ---
        vg_url_name = game_name.replace(' ', '+')
        vg_url = f"http://www.vgchartz.com/games/games.php?name={vg_url_name}"
        vg_response = requests.get(vg_url, timeout=10)
        if vg_response.status_code == 200:
            vg_soup = BeautifulSoup(vg_response.text, 'html.parser')
            table = vg_soup.find('table', {'class': 'chart'})
            if table:
                first_row = table.find('tr')
                if first_row:
                    cells = first_row.find_all('td')
                    if len(cells) >= 10:
                        # Use regex to match region labels flexibly
                        region_text = [c.text.strip().lower() for c in cells]
                        for i, text in enumerate(region_text):
                            if re.search(r'north\s*america|na', text):
                                info['NA_Sales'] = float(cells[i].text.strip())
                            elif re.search(r'europe|eu', text):
                                info['EU_Sales'] = float(cells[i].text.strip())
                            elif re.search(r'japan|jp', text):
                                info['JP_Sales'] = float(cells[i].text.strip())
                            elif re.search(r'other', text):
                                info['Other_Sales'] = float(cells[i].text.strip())
                            elif re.search(r'global', text):
                                info['Global_Sales'] = float(cells[i].text.strip())

    except Exception as e:
        print(f"Error scraping {game_name}: {e}")

    return info
    
missing_data = vgsales[
    vgsales["Publisher"].isna() |
    vgsales["Year"].isna() |
    (vgsales[sales_cols].isna() | (vgsales[sales_cols] == 0)).any(axis=1)
]

for idx, row in missing_data.iterrows():
    game_name = row["Name"]
    print(f"Scraping {game_name}...")
    info = get_game_info(game_name)
    
    for key, value in info.items():
        if pd.isna(vgsales.at[idx, key]) and value is not None:
            vgsales.at[idx, key] = value
    
    time.sleep(1)  # avoid being blocked

vgsales.to_csv("vgsales_scraped_full.csv", index=False)
print("✅ Updated dataset saved as vgsales_scraped_full.csv")
```

#### Is there a correlation between publishers and video game sales?

First let's take a look at our publishers and which are the top publishers in Global Sales. 

```python
publisher_sales = vgsales_cleaned.groupby('Publisher')['Global_Sales'].sum().sort_values(ascending=False)
publisher_sales.head(10)
```

```python
top_publishers = publisher_sales.head(10)

colors = plt.cm.rainbow(np.linspace(0, 1, len(top_publishers)))[::-1]

plt.figure(figsize=(10, 5))
top_publishers.plot(kind='bar', color=colors)
plt.title('Top 10 Publishers by Global Sales')
plt.ylabel('Global Sales (Millions)')
plt.xlabel('Publisher')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

Let's see if there is any correlation between publisher and sales.

```python
publisher_dummies = pd.get_dummies(vgsales_cleaned['Publisher'])

# Add sales column
publisher_dummies['Global_Sales'] = vgsales_cleaned['Global_Sales']

# Compute correlations
publisher_corr = publisher_dummies.corr()['Global_Sales'].sort_values(ascending=False)

publisher_corr.head(10)
```

Nintendo seems to be the only publisher that shows a moderate positive correlation at 0.27 in global sales, while the others have very weak correlations all being less than 0.06.

Let's run a statistical significance test to further show the difference.

```python
nintendo_sales = vgsales_cleaned.loc[vgsales_cleaned['Publisher'] == 'Nintendo', 'Global_Sales']
other_sales = vgsales_cleaned.loc[vgsales_cleaned['Publisher'] != 'Nintendo', 'Global_Sales']

t_stat, p_val = ttest_ind(nintendo_sales, other_sales, equal_var=False)  # Welch's t-test

print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_val:.6f}")

if p_val < 0.05:
    print("✅ The difference in sales between Nintendo and other publishers is statistically significant.")
else:
    print("❌ No statistically significant difference found.")
```

The visualization is a bit difficult to dechipher. 

Let's make seperate graphs showing the global sales by platform by decades.

Each decade will only show the top 10 sold platforms to be able to dechiper the vizualizations better. 

```python
decades = [
    (1980, 1989),
    (1990, 1999),
    (2000, 2010),
    (2010, 2016)
]

# Group sales by Year and Platform
platform_sales = vgsales_cleaned.groupby(['Year', 'Platform'])['Global_Sales'].sum().reset_index()

# Plot each decade separately
for start, end in decades:
    decade_df = platform_sales[(platform_sales['Year'] >= start) & (platform_sales['Year'] <= end)]
    
    # Find top 10 platforms by total sales in the decade
    top_platforms = (decade_df.groupby('Platform')['Global_Sales']
                     .sum()
                     .sort_values(ascending=False)
                     .head(10)
                     .index)
    
    top_decade_df = decade_df[decade_df['Platform'].isin(top_platforms)]
    
    plt.figure(figsize=(12, 6))
    sns.lineplot(data=top_decade_df, x='Year', y='Global_Sales', hue='Platform', marker='o')
    
    plt.title(f"Top 10 Platforms Global Sales ({start}-{end})", fontsize=14)
    plt.xlabel("Year", fontsize=12)
    plt.ylabel("Global Sales (Millions)", fontsize=12)
    plt.legend(title='Platform', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.show()
```

We can also show the top 10 platforms throughout all the years. 

```python
# Sum total global sales by platform over all years
top_platforms = (
    vgsales_cleaned.groupby('Platform')['Global_Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .index
)

# Filter data to include only top 10 platforms
top_platform_df = vgsales_cleaned[vgsales_cleaned['Platform'].isin(top_platforms)]

# Group by Year and Platform to sum sales per year
top_platform_sales = (
    top_platform_df.groupby(['Year', 'Platform'])['Global_Sales']
    .sum()
    .reset_index()
)

# Plot
plt.figure(figsize=(14, 7))
sns.lineplot(data=top_platform_sales, x='Year', y='Global_Sales', hue='Platform', marker='o')

plt.title("Global Sales Over Time for Top 10 Platforms", fontsize=14)
plt.xlabel("Year", fontsize=12)
plt.ylabel("Global Sales (Millions)", fontsize=12)
plt.legend(title='Platform', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Sum total global sales per platform
platform_totals = vgsales_cleaned.groupby('Platform')['Global_Sales'].sum()

# Get the top 3 platforms
top3_platforms = platform_totals.sort_values(ascending=False).head(3)

print("🏆 Top 3 Platforms of All Time:")
for i, (platform, sales) in enumerate(top3_platforms.items(), start=1):
    print(f"{i}. {platform} - {sales:.2f} million sales")
```

## Concluding Thoughts and Observations

This is more likely due to Japans historically and culturaly prefrence to swords due to a rich tradition of swordmanship and samurais. Famously, the 2008 recession could have had a major impact on consumers. Even though 2008 was the highest year in sales, the following years could have been directly impacted from the recession. Check out this article written in the beginning of 2010 disscussing this, written by Dean Takahashi on <a href="https://gamesbeat.com/that-screeching-sound-u-s-video-game-industry-sales-decline-in-2009/" target=blank>gamesbeat.com</a>.

Or maybe the rate of video games coming out could have slowed. 

Another possibility could have been phone games had started becoming more of a thing as cell phones began to advanced causing sales in typical platforms to drop.
